```python
from pathlib import Path
import pandas as pd

folder = Path("../data/can-ml/can-ml/2017-subaru-forester/090726/merged")

dfs = []
for csv_file in folder.glob("*.csv"):
    temp = pd.read_csv(csv_file)
    temp["source_file"] = csv_file.name
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)
```

```python
df = df.sort_values("timestamp", ascending=True).reset_index(drop=True)
print("Length of df: ", len(df))
print(df.head())
```

```python
combined = df[df["source_file"] == "combined.csv"]
print("Length of combined: ", len(combined))
combined_normal = combined[combined["attack"] == 0]
print("Length of combined_normal: ", len(combined_normal))
print("Ratio of combined_normal: ", len(combined_normal) / len(combined))
combined_attack = combined[combined["attack"] == 1]
print("Length of combined_attack: ", len(combined_attack))
print("Ratio of combined_attack: ", len(combined_attack) / len(combined))

dos = df[df["source_file"] == "DoS.csv"]
print("Length of dos: ", len(dos))
dos_normal = dos[dos["attack"] == 0]
print("Length of dos_normal: ", len(dos_normal))
print("Ratio of dos_normal: ", len(dos_normal) / len(dos))
dos_attack = dos[dos["attack"] == 1]
print("Length of dos_attack: ", len(dos_attack))
print("Ratio of dos_attack: ", len(dos_attack) / len(dos))

fuzzing = df[df["source_file"] == "fuzzing.csv"]
print("Length of fuzzing: ", len(fuzzing))
fuzzing_normal = fuzzing[fuzzing["attack"] == 0]
print("Length of fuzzing_normal: ", len(fuzzing_normal))
print("Ratio of fuzzing_normal: ", len(fuzzing_normal) / len(fuzzing))
fuzzing_attack = fuzzing[fuzzing["attack"] == 1]
print("Length of fuzzing_attack: ", len(fuzzing_attack))
print("Ratio of fuzzing_attack: ", len(fuzzing_attack) / len(fuzzing))

gear = df[df["source_file"] == "gear.csv"]
print("Length of gear: ", len(gear))
gear_normal = gear[gear["attack"] == 0]
print("Length of gear_normal: ", len(gear_normal))
print("Ratio of gear_normal: ", len(gear_normal) / len(gear))
gear_attack = gear[gear["attack"] == 1]
print("Length of gear_attack: ", len(gear_attack))
print("Ratio of gear_attack: ", len(gear_attack) / len(gear))


```

```python
interval = df[df["source_file"] == "interval.csv"]
print("Length of interval: ", len(interval))
print("Ratio of interval: ", len(interval) / len(df))
interval_normal = interval[interval["attack"] == 0]
print("Length of interval_normal: ", len(interval_normal))
print("Ratio of interval_normal: ", len(interval_normal) / len(interval))
interval_attack = interval[interval["attack"] == 1]
print("Length of interval_attack: ", len(interval_attack))
print("Ratio of interval_attack: ", len(interval_attack) / len(interval))

rpm = df[df["source_file"] == "rpm.csv"]
print("Length of rpm: ", len(rpm))
rpm_normal = rpm[rpm["attack"] == 0]
print("Length of rpm_normal: ", len(rpm_normal))
print("Ratio of rpm_normal: ", len(rpm_normal) / len(rpm))
rpm_attack = rpm[rpm["attack"] == 1]
print("Length of rpm_attack: ", len(rpm_attack))
print("Ratio of rpm_attack: ", len(rpm_attack) / len(rpm))
```

```python
speed = df[df["source_file"] == "speed.csv"]
print("Length of speed: ", len(speed))
speed_normal = speed[speed["attack"] == 0]
print("Length of speed_normal: ", len(speed_normal))
print("Ratio of speed_normal: ", len(speed_normal) / len(speed))
speed_attack = speed[speed["attack"] == 1]
print("Length of speed_attack: ", len(speed_attack))
print("Ratio of speed_attack: ", len(speed_attack) / len(speed))

standstill = df[df["source_file"] == "standstill.csv"]
print("Length of standstill: ", len(standstill))
standstill_normal = standstill[standstill["attack"] == 0]
print("Length of standstill_normal: ", len(standstill_normal))
print("Ratio of standstill_normal: ", len(standstill_normal) / len(standstill))
standstill_attack = standstill[standstill["attack"] == 1]
print("Length of standstill_attack: ", len(standstill_attack))
print("Ratio of standstill_attack: ", len(standstill_attack) / len(standstill))

systematic = df[df["source_file"] == "systematic.csv"]
print("Length of systematic: ", len(systematic))
systematic_normal = systematic[systematic["attack"] == 0]
print("Length of systematic_normal: ", len(systematic_normal))
print("Ratio of systematic_normal: ", len(systematic_normal) / len(systematic))
systematic_attack = systematic[systematic["attack"] == 1]
print("Length of systematic_attack: ", len(systematic_attack))
print("Ratio of systematic_attack: ", len(systematic_attack) / len(systematic))
```
