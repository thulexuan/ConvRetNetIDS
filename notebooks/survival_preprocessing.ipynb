```python
import os
import pandas as pd
```

```python
column_names = ["timestamp", "arbitration_id", "dlc", "data0", "data1", "data2", "data3",
                 "data4", "data5", "data6", "data7", "attack"]
```

```python
def fill_flag(sample):
    if not isinstance(sample['attack'], str):
        col = 'data' + str(sample['dlc'])
        sample['attack'], sample[col] = sample[col], sample['attack']
    return sample
```

```python
flooding_path = '../data/survival_dataset/Sonata/Flooding_dataset_SONATA.txt'
fuzzy_path = '../data/survival_dataset/Sonata/Fuzzy_dataset_SONATA.txt'
malfunction_path = '../data/survival_dataset/Sonata/Malfunction_dataset_SONATA.txt'
```

```python
flooding_df = pd.read_csv(flooding_path, names=column_names, sep=',', header=None)
print("Length of flooding file: ", len(flooding_df))
flooding_df = flooding_df.apply(fill_flag, axis=1)
print(flooding_df.head())
```

```python
print(len(flooding_df[flooding_df['attack'] == 'R']))
print(len(flooding_df[flooding_df['attack'] != 'R']))
```

```python
fuzzy_df = pd.read_csv(fuzzy_path, names=column_names, sep=',', header=None)
print("Length of fuzzy file: ", len(fuzzy_df))
fuzzy_df = fuzzy_df.apply(fill_flag, axis=1)
print(fuzzy_df.head())
```

```python
print(len(fuzzy_df[fuzzy_df['attack'] == 'R']))
print(len(fuzzy_df[fuzzy_df['attack'] != 'R']))
```

```python
fuzzy_df[fuzzy_df['dlc'] < 8].head()
```

```python
malfunction_df = pd.read_csv(malfunction_path, names=column_names, sep=',', header=None)
print("Length of malfunction file: ", len(malfunction_df))
print(malfunction_df.head())
malfunction_df = malfunction_df.apply(fill_flag, axis=1)
print(malfunction_df.head())
```

```python
print(len(malfunction_df[malfunction_df['attack'] == 'R']))
print(len(malfunction_df[malfunction_df['attack'] != 'R']))
```

```python
malfunction_df[malfunction_df['dlc'] < 8].head()
```

```python
# convert id from hex to dec
flooding_df['arbitration_id'] = flooding_df['arbitration_id'].apply(lambda x: int(x, 16) if isinstance(x, str) else x)
fuzzy_df['arbitration_id'] = fuzzy_df['arbitration_id'].apply(lambda x: int(x, 16) if isinstance(x, str) else x)
malfunction_df['arbitration_id'] = malfunction_df['arbitration_id'].apply(lambda x: int(x, 16) if isinstance(x, str) else x)
```

```python
print(flooding_df.head(1))
print(fuzzy_df.head(1))
print(malfunction_df.head(1))
```

```python
print("Filling NaN values with '00': ")
flooding_df = flooding_df.fillna("00")
fuzzy_df = fuzzy_df.fillna("00")
malfunction_df = malfunction_df.fillna("00")

print("Filling NaN values with '00' done.")
print(flooding_df[flooding_df['dlc'] < 8].head(1))
print(fuzzy_df[fuzzy_df['dlc'] < 8].head(1))
print(malfunction_df[malfunction_df['dlc'] < 8].head(1))
```

```python
byte_columns = [f"data{i}" for i in range(8)]
flooding_df['data_field'] = flooding_df[byte_columns].astype(str).agg(''.join, axis=1)
fuzzy_df['data_field'] = fuzzy_df[byte_columns].astype(str).agg(''.join, axis=1)
malfunction_df['data_field'] = malfunction_df[byte_columns].astype(str).agg(''.join, axis=1)

print(flooding_df['data_field'].head(1))
print(fuzzy_df['data_field'].head(1))
print(malfunction_df['data_field'].head(1))
```

```python
print("Upper case data_field column: ")
flooding_df['data_field'] = flooding_df['data_field'].str.upper()
fuzzy_df['data_field'] = fuzzy_df['data_field'].str.upper()
malfunction_df['data_field'] = malfunction_df['data_field'].str.upper()

print("Showing fuzzy: ")
print(flooding_df["data_field"].head())
print("Showing gear: ")
print(fuzzy_df["data_field"].head())
print("Showing rpm: ")
print(malfunction_df["data_field"].head())
```

```python
print("Changing value of attack column with R -> 0, T --> 1")
flooding_df['attack'] = flooding_df['attack'].map({'R': 0, 'T': 1})
fuzzy_df['attack'] = fuzzy_df['attack'].map({'R': 0, 'T': 1})
malfunction_df['attack'] = malfunction_df['attack'].map({'R': 0, 'T': 1})
print("After changing value of attack column with R -> 0, T --> 1: ")
print(flooding_df["attack"].head())
print(fuzzy_df["attack"].head())
print(malfunction_df["attack"].head())
```

```python
print("Collecting necessary columns for the next preprocessing step:  ")
selected_fuzzy_df = fuzzy_df[['timestamp', 'arbitration_id', 'data_field', 'attack']]
selected_flooding_df = flooding_df[['timestamp', 'arbitration_id', 'data_field', 'attack']]
selected_malfunction_df = malfunction_df[['timestamp', 'arbitration_id', 'data_field', 'attack']]

print("Checking before putting into the next preprocessing step file: ")
print(selected_fuzzy_df.head())
print(selected_flooding_df.head())
print(selected_malfunction_df.head())
```

```python
output_file_fuzzy = '../data/survival_dataset/Sonata-preprocessed/first_preprocessing/fuzzy.csv'
output_file_flooding = '../data/survival_dataset/Sonata-preprocessed/first_preprocessing/flooding.csv'
output_file_malfunction = '../data/survival_dataset/Sonata-preprocessed/first_preprocessing/malfunction.csv'

selected_fuzzy_df.to_csv(output_file_fuzzy, index=False)
selected_flooding_df.to_csv(output_file_flooding, index=False)
selected_malfunction_df.to_csv(output_file_malfunction, index=False)
```
