```python
import os
import pandas as pd
```

```python
column_names = ["timestamp", "arbitration_id", "dlc", "data0", "data1", "data2", "data3",
                 "data4", "data5", "data6", "data7", "attack"]
```

```python
def fill_flag(sample):
    if not isinstance(sample['attack'], str):
        col = 'data' + str(sample['dlc'])
        sample['attack'], sample[col] = sample[col], sample['attack']
    return sample
```

### preprocessing 

```python
# Define the directory path
dos_path = '../data/car-hacking/DoS_dataset.csv'
fuzzy_path = '../data/car-hacking/Fuzzy_dataset.csv'
gear_path = '../data/car-hacking/gear_dataset.csv'
rpm_path = '../data/car-hacking/RPM_dataset.csv'

dos_df = pd.read_csv(dos_path, header=None, names=column_names)
fuzzy_df = pd.read_csv(fuzzy_path, header=None, names=column_names)
gear_df = pd.read_csv(gear_path, header=None, names=column_names)
rpm_df = pd.read_csv(rpm_path, header=None, names=column_names)
```

```python
print("Length of dos file: ", len(dos_df))
print("Length of fuzzy file: ", len(fuzzy_df))
print("Length of gear file: ", len(gear_df))
print("Length of rpm file: ", len(rpm_df))
```

```python
print("Show fuzzy dataframe: ")
print(fuzzy_df.head())
print("Show gear dataframe: ")
print(gear_df.head())
print("Show rpm dataframe: ")
print(rpm_df.head())
```

```python
print("Checking for rows with dlc == 2 in fuzzy dataframe: ")
print(fuzzy_df[fuzzy_df['dlc'] == 2])
print("Checking for rows with dlc == 2 in gear dataframe: ")
print(gear_df[gear_df['dlc'] == 2])
print("Checking for rows with dlc == 2 in rpm dataframe: ")
print(rpm_df[rpm_df['dlc'] == 2])
```

```python
print("Filling flag: ")
ds_df = dos_df.apply(fill_flag, axis=1)
fuzzy_df = fuzzy_df.apply(fill_flag, axis=1)
gear_df = gear_df.apply(fill_flag, axis=1)
rpm_df = rpm_df.apply(fill_flag, axis=1)
```

```python
normal_dos = dos_df[dos_df['attack'] == 'R']
print("Length of normal dos file: ", len(normal_dos))
dos_attack = dos_df[dos_df['attack'] == 'T']
print("Length of dos attack file: ", len(dos_attack))

normal_fuzzy = fuzzy_df[fuzzy_df['attack'] == 'R']
print("Length of normal fuzzy file: ", len(normal_fuzzy))
normal_gear = gear_df[gear_df['attack'] == 'R']
print("Length of normal gear file: ", len(normal_gear))
normal_rpm = rpm_df[rpm_df['attack'] == 'R']
print("Length of normal rpm file: ", len(normal_rpm))
fuzzy_attack = fuzzy_df[fuzzy_df['attack'] == 'T']
print("Length of fuzzy attack file: ", len(fuzzy_attack))
gear_attack = gear_df[gear_df['attack'] == 'T']
print("Length of gear attack file: ", len(gear_attack))
rpm_attack = rpm_df[rpm_df['attack'] == 'T']
print("Length of rpm attack file: ", len(rpm_attack))
```

```python
print("After filling flag: ")
print(fuzzy_df[fuzzy_df['dlc'] == 2])
print(gear_df[gear_df['dlc'] == 2])
print(rpm_df[rpm_df['dlc'] == 2])
```

```python
print("Filling NaN values with '00': ")
fuzzy_df = fuzzy_df.fillna("00")
gear_df = gear_df.fillna("00")
rpm_df = rpm_df.fillna("00")
```

```python
print("After filling NaN columns: ")
print(fuzzy_df[fuzzy_df['dlc'] == 2])
print(gear_df[gear_df['dlc'] == 2])
print(rpm_df[rpm_df['dlc'] == 2])
```

```python
byte_columns = [f"data{i}" for i in range(8)]
fuzzy_df['data_field'] = fuzzy_df[byte_columns].astype(str).agg(''.join, axis=1)
gear_df['data_field'] = gear_df[byte_columns].astype(str).agg(''.join, axis=1)
rpm_df['data_field'] = rpm_df[byte_columns].astype(str).agg(''.join, axis=1)
```

```python
print("After concating 8 data field columns: ")
print(fuzzy_df["data_field"].head())
print(gear_df["data_field"].head())
print(rpm_df["data_field"].head())
```

```python
print("Upper case data_field column: ")
fuzzy_df['data_field'] = fuzzy_df['data_field'].str.upper()
gear_df['data_field'] = gear_df['data_field'].str.upper()
rpm_df['data_field'] = rpm_df['data_field'].str.upper()

print("Showing fuzzy: ")
print(fuzzy_df["data_field"].head())
print("Showing gear: ")
print(gear_df["data_field"].head())
print("Showing rpm: ")
print(rpm_df["data_field"].head())
```

```python
print("Converting arbitration_id from hex to int: ")
fuzzy_df['arbitration_id'] = fuzzy_df['arbitration_id'].apply(lambda x: int(x, 16) if isinstance(x, str) else x)
gear_df['arbitration_id'] = gear_df['arbitration_id'].apply(lambda x: int(x, 16) if isinstance(x, str) else x)
rpm_df['arbitration_id'] = rpm_df['arbitration_id'].apply(lambda x: int(x, 16) if isinstance(x, str) else x)
```

```python
print("After converting can_id from hex to dec: ")
print(fuzzy_df["arbitration_id"].head())
print(gear_df["arbitration_id"].head())
print(rpm_df["arbitration_id"].head())
```

```python
print("Changing value of attack column with R -> 0, T --> 1")
fuzzy_df['attack'] = fuzzy_df['attack'].map({'R': 0, 'T': 1})
gear_df['attack'] = gear_df['attack'].map({'R': 0, 'T': 1})
rpm_df['attack'] = rpm_df['attack'].map({'R': 0, 'T': 1})
print("After changing value of attack column with R -> 0, T --> 1: ")
print(fuzzy_df["attack"].head())
print(gear_df["attack"].head())
print(rpm_df["attack"].head())
```

```python
print("Collecting necessary columns for the next preprocessing step:  ")
selected_fuzzy_df = fuzzy_df[['timestamp', 'arbitration_id', 'data_field', 'attack']]
selected_gear_df = gear_df[['timestamp', 'arbitration_id', 'data_field', 'attack']]
selected_rpm_df = rpm_df[['timestamp', 'arbitration_id', 'data_field', 'attack']]

print("Checking before putting into the next preprocessing step file: ")
print(selected_fuzzy_df.head())
print(selected_gear_df.head())
print(selected_rpm_df.head())
```

```python
output_file_fuzzy = '../data/car-hacking/first_preprocessing/fuzzy.csv'
output_file_gear = '../data/car-hacking/first_preprocessing/gear.csv'
output_file_rpm = '../data/car-hacking/first_preprocessing/rpm.csv'

selected_fuzzy_df.to_csv(output_file_fuzzy, index=False)
selected_gear_df.to_csv(output_file_gear, index=False)
selected_rpm_df.to_csv(output_file_rpm, index=False)
```

```python
import os
import pandas as pd

# Directory path containing CSV files
# /home/hieutt/UniCon/data/can-ml/2016-chevrolet-silverado/merged
input_dir = '../data/can-ml/can-ml/2016-chevrolet-silverado/merged'  # Change the path if needed

# List of CSV files in the directory
files = [f for f in os.listdir(input_dir) if f.endswith('.csv')]

# Function to check each file
def check_data(file_path):
    print(f"\nChecking file: {file_path}")
    
    # Read the file
    df = pd.read_csv(file_path)
    
    # Display the shape of the DataFrame
    print("Shape:", df.shape)
    
    # Display the first 5 rows
    print("Head:")
    display(df.head())
    
    # Check for NaN values
    nan_count = df.isna().sum().sum()
    if nan_count > 0:
        print("Warning: There are NaN values in the data.")
        print(df.isna().sum())
    else:
        print("No NaN values found.")
    
    # Check data types
    print("\nData types:")
    print(df.dtypes)
    
    # Check label distribution if 'label' column exists
    if 'label' in df.columns:
        print("\nLabel distribution:")
        print(df['label'].value_counts())
    
    # Descriptive statistics for numeric columns
    print("\nDescriptive statistics for numeric columns:")
    print(df.describe())
    
# Iterate over each file and check
for file in files:
    check_data(os.path.join(input_dir, file))
```
